```python
class ReflexTemperatureAgent:
  def act(self,temp):
    if temp>30:
      return "TURN ON FAN"
    elif temp<20:
      return "TURN ON HEATER"
    else:
      return "MAINTAIN CURRENT TEMPERATURE"
```

```python
agent=ReflexTemperatureAgent()
```

```python
temp=[10,30,45,15,40]
```

```python
for t in temp:
  action=agent.act(t);
  print("Temperature:",t)
  print("Action",action)
```
