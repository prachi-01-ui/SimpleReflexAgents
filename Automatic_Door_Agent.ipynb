```python
class AutomaticDoorAgent:
  def act(self,Person):
    if Person==True:
      return "OPEN THE DOOR"
    else:
      return "CLOSE THE DOOR"
```

```python
agent=AutomaticDoorAgent()
```

```python
Person=[True,True,False,False,True]
```

```python
for p in Person:
  action=agent.act(p);
  print("Person detected:",p)
  print("Action",action)
```
